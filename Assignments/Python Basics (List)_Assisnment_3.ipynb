## Question: 03
    Find and display the largest number of a list without using built-in function max(). Your program
    should ask the user to input values in list from keyboard.

```python
num_li = []
for item in range(5):
    num_li_items = int( input(f"Enter {item+1} item of list :"))
    num_li.append(num_li_items)
```

```python
largest = num_li[0]
for num in num_li:
    if num > largest:
        largest = num
print(f"Largest number is {largest} and it's on index {num_li.index(largest)}")
```

## Question: 06
    Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It
    should print the date in the form March 12, 2021.

```python
months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

date = input("Enter the Date (mm/dd/yyyy) eg: 12/28/2001: ")

mon, day, year = date.split('/')

mon_name = months[int(mon) - 1]

print(f"{mon_name} {int(day)}, {year}")
```

## Question: 08
     Find the sum of each row of matrix of size m x n. For example for the following matrix output will be
    like this :

    matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
    ]
        Sum of row 1 = 32
        Sum of row 2 = 31
        Sum of row 3 = 63

```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
    ]

for i, row in enumerate(matrix, start=1):
    row_sum = sum(row)
    print(f"Sum of row {i} = {row_sum}")
```

## Question:09
    Write a program to add two matrices of size n x m.

```python
A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8, 9],
    [1, 2, 3]
]

result = [
    [0, 0, 0],
    [0, 0, 0]
]

for i in range(len(A)):          
    for j in range(len(A[0])):  
        result[i][j] = A[i][j] + B[i][j]

print("Matrix A + Matrix B:")
for row in result:
    print(row)
```

## Question:10
    Write a program to multiply two matrices.

```python
A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8],
    [9, 1],
    [2, 3]
]

result = [
    [0, 0],
    [0, 0]
]


for i in range(len(A)):         
    for j in range(len(B[0])):     
        for k in range(len(B)):   
            result[i][j] += A[i][k] * B[k][j]
    
print("Matrix Multiplication Result:")
for row in result:
    print(row)
```
