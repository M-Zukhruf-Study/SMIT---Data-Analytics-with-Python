### Question: 03
    Write a program in which use all the operators we can use in Python.

```python
num_1 = float(input("Enter the first number: "))

oprt = input("Enter the operator (+, -, *, /, <=, >=, ==,%): ")

num_2 = float(input("Enter the second number: "))
```

```python
if (oprt == "+"):
    print ( num_1 + num_2)
elif(oprt == "-"):
    print (num_1 - num_2)
elif ( oprt =="/"):
    print(num_1 / num_2)
elif ( oprt =="*"):
    print(num_1 * num_2)
elif ( oprt =="<="):
    print(num_1 <= num_2)
elif ( oprt ==">="):
    print(num_1 >= num_2) 
elif ( oprt =="=="):
    print(num_1 == num_2)
elif ( oprt =="%"):
    print(num_1 % num_2)
else:
    print ("You've added something wrong")
```

### Question: 04
     Completes the following steps of small task:
        Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
        Mention Variable of Total Marks and assign 300 to it
        Calculate Percentage
        

```python
eng =int(input("Enter English marks :"))
math =int(input("Enter Mathmathics marks"))
isl =int(input("Enter Islamiat marks"))
```

```python
def check_eng(eng):
    if (eng <= 100):
        return eng
    else:
        return ("Kindly enter valid numbers of English")
    
def check_math(math):
    if (math <= 100):
        return math
    else:
        return ("Kindly enter valid numbers of Mathmatics")
    
def check_isl(isl):
    if (isl <= 100):
        return isl
    else:
        return ("Kindly enter valid numbers of Islamiat")
```

```python
english=check_eng(eng)
mathmatics=check_math(math)
islamiat=check_isl(isl)

if english != eng:
    print(english)
if islamiat != isl:
    print(islamiat)
if mathmatics != math:
    print(mathmatics)
```

```python
sum_marks= int(english+mathmatics+islamiat)
```

```python
total_marks=300
percentage=(sum_marks/total_marks)*100
```

```python
print(f"Your English marks are {english}\nYour Islamiat marks are {islamiat}\nYour Mathmatics marks are {mathmatics}\nYour Total marks are {sum_marks}\nYour percentage is {percentage}")
```
