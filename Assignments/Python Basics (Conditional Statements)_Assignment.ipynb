## Question: 01

    A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years. Ask user for their salary and year of service and print the net bonus amount.
    

```python
user_name = input("Enter your name")
user_salary = int(input("Enter your Current Salary in Digits :"))
yo_service = int(input("Enter your Year Of Service in Digits :"))
```

```python
def bonus_calculator(salary, name, service):
    if yo_service > 5:
        salary_cal = user_salary * 0.05
        final_slary = user_salary + salary_cal
        return f"Hey! {name}, Your total years of service is {service}, current salary is {salary} and total bonus amount is {salary_cal}, so total salary is {final_slary}."
    
    else:
        return f"Hey! {name}, we are really sorry you didn't get any bonus. Your service years didn't match criteria. Current service year's {service}"
    
    
    
    
bonus_calculator(salary=user_salary, name=user_name, service=yo_service)
```

## Question: 06
    Write a program to calculate the electricity bill (accept number of unit from user) according to the following criteria : Unit Price

```python
def unit_cal(unit):
    if unit >= 0 and unit <=100:
        bill = 0 * unit
        return f"Your bill is RS{bill}"
    
    elif unit > 100 and unit <=300:
            bill = 5 * unit
            return f"Your bill is RS{bill}"
        
    elif unit > 300:
                bill = 10 * unit
                return f"Your bill is RS : {bill}"
    else:
        return "Invalid! input"
```

```python
print(unit_cal(int(input("Enter the units :"))))
```

## Question: 11
    
    A school has following rules for grading system:

    a. Below 25 - F

    b. 25 to 45 - E

    c. 45 to 50 - D

    d. 50 to 60 - C

    e. 60 to 80 - B

    f. Above 80 - A

    Ask user to enter marks and print the corresponding grade.

```python
def grade_cal(marks):
    if marks < 25 :
        return "Your Grade is F"
    elif marks >= 25 and marks < 45 :
        return "Your Grade is E"
    elif marks >= 45 and marks < 50 :
        return "Your Grade is D"
    elif marks >= 50 and marks < 60 :
        return "Your Grade is C"
    elif marks >= 60 and marks < 80 :
        return "Your Grade is B"
    elif marks >= 80:
        return "Your Grade is A"
```

```python
marks = (int(input("Enter the Marks :")))
grade_cal(marks)
```

## Question:12
    Take input of age of 3 people by user and determine oldest and youngest among them.

```python
def age_cal(a,b,c):
    old = max(a,b,c)
    young = min (a,b,c)
    return f"The oldest is {old} and youngest is {young}"
```

```python
a = (int(input("Enter the AGE :")))
b = (int(input("Enter the AGE :")))
c = (int(input("Enter the AGE :")))
```

```python
age_cal(a=a,b=b,c=c)
```
