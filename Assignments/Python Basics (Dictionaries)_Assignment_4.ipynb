## Question: 03
    Loop Through Dictionary
        person = {
        "name": "Ali",
        "age": 25,
        "city": "Karachi"
        }
    Print keys and values using loop

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(f"{key}: {value}")
```

## Question: 04
    Count Frequency of Numbers
        numbers = [1,2,2,3,3,3,4,4,4,4]
        Create frequency dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
freq = {}

for num in numbers:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print("Frequency Dictionary:", freq)
```

## Question: 05
     Word Counter
        Take sentence input and count word frequency

```python
sentence = input("Sentence enter karein: ")
words = sentence.split()
word_count = {}

for word in words:
    word = word.lower()
    
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print("Word Frequency:", word_count)
```

## Question: 06
     Student Marks & Average
        marks = {"Math":80, "English":70, "Physics":90}
        Calculate total and average

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total_marks = sum(marks.values())
average_marks = total_marks / len(marks)

print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

## Question: 07
     Highest Scoring Student
        students = {
        "Ali":85,
        "Sara":92,
        "Ahmed":78,
        "Zara":95
        }
    Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = max(students, key=students.get)
highest_score = students[highest_student]

print("Highest Scoring Student:", highest_student)
print("Score:", highest_score)
```
